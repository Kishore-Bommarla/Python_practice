```python
class Solution:
    def pattern1(self, N):
        for i in range(N):
            for j in range(N):
                print("*",end="")
            print()
# Driver code
sol = Solution()
N = int(input())  
sol.pattern1(N) 
```

```python
class Solution:
    def pattern6(self, N):
        for i in range(1, N + 1):

            # spaces
            for s in range(N - i):
                print(" ", end="")

            # stars
            for j in range(2*i - 1):
                print("*", end="")

            print()

# Driver code
sol = Solution()
N = int(input())
sol.pattern6(N)
```

```python
class Solution:
    def pattern7(self, N):
        for i in range(1,N+1):
            for j in range(N-i):
                print(" ", end="")
            for k in range(2*i-1):
                print("*",end="")
            for l in range(N-i):
                 print(" ", end="")
            print()

# Driver code
sol = Solution()
N = int(input())
sol.pattern7(N)
```

```python
class Solution:
    def pattern8(self, N):
        for i in range(1,N+1):
            for j in range(i-1):
                print(" ", end="")
            for k in range(2 * N - (2 * i + 1)):
                print("*",end="")
            for l in range(i-1):
                 print(" ", end="")
            print()

# Driver code
sol = Solution()
N = int(input())
sol.pattern8(N)
```

```python
class Solution:
    def pattern7(self, N):
        # Upper pyramid
        for i in range(1, N + 1):
            for j in range(N - i):
                print(" ", end="")
            for k in range(2 * i - 1):
                print("*", end="")
            print()

        # Middle line again
        for k in range(2 * N - 1):
            print("*", end="")
        print()

    def pattern8(self, N):
        # Lower pyramid
        for i in range(N - 1, 0, -1):
            for j in range(N - i):
                print(" ", end="")
            for k in range(2 * i - 1):
                print("*", end="")
            print()


# Driver code
sol = Solution()
N = int(input())
sol.pattern7(N)
sol.pattern8(N)

```
